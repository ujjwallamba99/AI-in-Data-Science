#Data visualization with matplotlib

##3.1 Line Chart

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
marks = [98,60,52,72,80]
noh = [10,6,5,7,8]

# Create a figure with a specific size (width, height)
plt.figure(figsize=(5,6))
# Plot the data with 'marks' on the x-axis and 'noh' on the y-axis.
# 'marker="s"' uses square markers and 'linestyle="-."' uses a dash-dot line style.
plt.plot(marks,noh,marker="s",linestyle="-.")

# Set the title of the plot
plt.title("Marks of Students",fontsize=14,color="blue")
# Set the label for the x-axis
plt.xlabel("Marks",fontsize=10,color="red")
# Set the label for the y-axis
plt.ylabel("No. of Study Hours",fontsize=10,color="green")


# Add a grid to the plot for better readability
plt.grid(True)
# Display the plot
plt.show()
```

```python
student={
    "marks":[20,22,25,30,35]
}
df= pd.DataFrame(student)

plt.figure(figsize=(6,5)) # Create a figure with a specific size (width, height)
plt.plot(df["marks"],marker="s",linestyle="-.") # Plot the 'marks' column from the DataFrame with square markers and a dash-dot line style

plt.title("Marks of Students",fontsize=14,color="blue")     # Set the title of the plot
plt.xlabel("Marks",fontsize=10,color="red") # Set the label for the x-axis
plt.ylabel("No. of Study Hours",fontsize=10,color="green") # Set the label for the y-axis


plt.grid(True) # Add a grid to the plot for better readability
plt.show() # Display the plot
```

```python
import pandas as pd
import matplotlib.pyplot as plt

student = {
    "marks": [20, 22, 25, 30, 35]
}

df = pd.DataFrame(student)

# FOR LOOP + RANGE + INDEX
print("Using for loop, range and index:")

for i in range(len(df)):
    print("Index:", i, "Marks:", df["marks"][i])


# ENUMERATE
print("\nUsing enumerate:")

for index, mark in enumerate(df["marks"]):
    print("Index:", index, "Marks:", mark)


#  WHILE LOOP
print("\nUsing while loop:")

i = 0

while i < len(df):
    print("Index:", i, "Marks:", df["marks"][i])
    i += 1


#  PLOTTING
plt.figure(figsize=(6, 5))

plt.plot(
    df.index,                 # Index
    df["marks"],
    marker="s",
    linestyle="-."
)

plt.title("Marks of Students", fontsize=14, color="blue")
plt.xlabel("Student Index", fontsize=10, color="red")
plt.ylabel("Marks", fontsize=10, color="green")

plt.grid(True)
plt.show()
```
