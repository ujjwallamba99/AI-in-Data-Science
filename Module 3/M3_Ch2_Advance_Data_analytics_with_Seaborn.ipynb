#Line plot

```python
#normal code
#add hue = "sex" as ivide data according to category
#line size using (size = 1,2,3) to change  line size
#change theme of lines according to auumn (palette = "rainbow")
#(markers = ["s","o"]) use (style = "sex") while using the category markers
#dashes = True or False line broke and simple acording to use
#legend by defaut visible (legend = True)

sns.lineplot(x = "bill_length_mm",y = "bill_depth_mm", data = df,hue = "sex",palette="rainbow",style ="sex",markers = ["s","o"],dashes = False,legend = False)

#Home Work
sns.set_theme(style="dark")  #styles: darkgrid (default), whitegrid, dark, white, and ticks



plt.show() # if we dont use show the shows warning but will work
```
