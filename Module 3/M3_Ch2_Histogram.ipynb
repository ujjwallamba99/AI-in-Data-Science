#3.2 Histogram

```python
import matplotlib.pyplot as plt
import pandas as pd
age = [-20,21,19,1,20,20,21,20,19,17,23,12,12,13,14,14,15,5,6,3,6]

plt.hist(age)

plt.title("Histogram",fontsize=14,color="blue")     # Set the title of the plot
plt.xlabel("Ages",fontsize=10,color="red") # Set the label for the x-axis
plt.ylabel("Frequency",fontsize=10,color="green") # Set the label for the y-axis


plt.grid(True) # Add a grid to the plot for better readability
plt.show() # Display the plot
```

```python
age = [10,20,20,20,20,22,22,30,25,45,50]

plt.figure(figsize=(6,5))

plt.hist(age,bins=4,edgecolor="k",color ="y",alpha=0.3) # Plotting a histogram with 4 bins, black edges, yellow bars, and 30% transparency

plt.title("Histogram",fontsize=14,color="blue")     # Set the title of the plot
plt.xlabel("Ages",fontsize=10,color="red") # Set the label for the x-axis
plt.ylabel("Frequency",fontsize=10,color="green") # Set the label for the y-axis


plt.grid(axis = "y") # Add a grid to the plot for better readability
plt.show() # Display the plot
```

Task

Create a line plot of the students' marks and perform the following:

1.Calculate the mean of the marks. \
2.Calculate the median of the marks.\
3.Plot the marks using plt.plot().\
4.Add a horizontal line for the mean using:

```python
#Add a horizontal line for the median using:
import pandas as pd
import matplotlib.pyplot as plt

# Data
student = {
    "marks": [20, 22, 25, 25, 28, 30, 32, 35, 40, 80]
}

df = pd.DataFrame(student)

# Calculate mean and median
mean = df["marks"].mean()
median = df["marks"].median()

print("Mean:", mean)
print("Median:", median)

# Histogram

plt.figure(figsize=(8, 5))

plt.hist(
    df["marks"],
    bins=5,
    edgecolor="black"
)

# Mean line
plt.axvline(
    mean,
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean:.2f}"
)

# Median line
plt.axvline(
    median,
    linestyle="-",
    linewidth=2,
    label=f"Median = {median:.2f}"
)

# Title and labels
plt.title("Distribution of Student Marks", fontsize=14)
plt.xlabel("Marks", fontsize=11)
plt.ylabel("Number of Students", fontsize=11)

# Grid
plt.grid(axis = "y", alpha=0.3)

# Legend
plt.legend()

# Show plot
plt.show()

```
