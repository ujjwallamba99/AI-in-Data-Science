'''Slicing Formula \

```python
s2 = "my name is ujjwal"
print(s2[2:5])  #print output from 2 to 4 because it takes index and index starts from zero
print(s2[:6])   #start to ixd 5
print(s2[10:])   #10 to last
print(s2[-2:-5])  #reverse slicing but not work because slicing go from left to right
print(s2[-5:-2]) #reverse slicing work
```
