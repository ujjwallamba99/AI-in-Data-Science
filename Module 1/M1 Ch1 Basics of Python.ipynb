#Relational Operator

```python
num1 = 1
num2 = 2
print(num1 == num2)  #is num1 is same as num 2
print(num1 != num2)  # is num 1 is not equal to num 2
print(num1 <= num2)  # is num 1 is lessthan equal to num 2
print(num1 >= num2)  #is num 1 is greaterthan equal to num 2
print(num1 < num2)  #is num 1 is less than to num 2
print(num1 > num2)  #is num 1 is greater than to num 2
```

#Logical Operators => and, or, not


```python
print(True and True)  #Both condition have to true
print(True and False)
print(False and True)
print(False and False)
```

```python
print(True or True)  #Atleast one condition have to true
print(True or False)
print(False or True)
print(False or False)
```
