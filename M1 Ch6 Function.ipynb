#Functions

```python
#example of creating function to get sum
# def : definition of fn
# Sum  : function name
# (a,b)  : parameter
# Sum(4,3)  : arguments


def Sum(a,b):
  Sum = a+b         #
  return Sum        ##  Body of fn
print(Sum(4,5))     #
```

```python
def juice(fruit):
  return fruit + " Juice"
juice("Apple")
```

Basic calculator using fn.

```python
def calculator(a,b):
  Sum = a+b
  Sub = a-b
  Mul = a*b
  Div = a/b
  Coef = a**b
  return Sum,Sub,Mul,Div,Coef

calculator(2,4)
```

#Avg using function

```python
def avg(a,b,c):
  avg = (a+b+c)/3
  return avg
print(avg(2,4,8))
```

```python
def avg(a,b,c=5):
  avg = (a+b+c)/3
  return avg
print(avg(2,4,6))     # works
```

(n! = n × (n-1)!) Formula of factorial

Q. WAP to find factorial on n number.

```python
# ques to find factorial on n number
#n = 5
n = 5
def factorial(n):
  fact = 1
  for i in range (1,n+1):
    fact = fact*i
  return fact
factorial(n)
```
