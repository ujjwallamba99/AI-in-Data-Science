Q. Write a code for hw to get input from user and and save user id in list.


```python
# Create an empty list to store user IDs
user_ids = []

# Ask the user how many IDs they want to enter
count = int(input("How many user IDs do you want to add? "))

# Loop to collect the input
for _ in range(count):
    uid = input("Enter user ID: ")
    user_ids.append(uid)  # Add the ID to the list

print("Saved User IDs:", user_ids)
```

##Set and Dict Practice Ques

1. Write a program to create a dictionary from two lists: one of keys and one of values.

---



```python
key = [ "a","b","c","d"]
value = [ "1","2","3","4"]

result = {}

for ele in range(len(key)):
  result[key[ele]] = value[ele]
print(result)
```

2. Merge two dictionaries into one

```python
l1 = {"a":1,"b":2}
l2 = {"c":3,"d":4}

l1.update(l2)


print(l1)
```

3. Write a program to sort a dictionary by its values.

```python
scores = {'aa': 22, 'bb': 55, 'cc': 34}

def get_value(item):
    return item[1]

sorted_dict = dict(sorted(scores.items(), key=get_value))

print(sorted_dict)
```

4. Given two sets, check if one set is a subset of another.

```python
set_a = {1, 2, 3}
set_b = {1, 2, 3, 4, 5}

is_subset = set_a.issubset(set_b)

print(is_subset)
```

5. Write a program to check whether two lists have at least one common element using sets.

```python
list1 = ['ram', 'sham', 'chaam']
list2 = ['sham', 'cham', 'dam']

common = set(list1) & set(list2)

if common:
    print("Common element exist!",common)
else:
    print("No common elements found.")
```
