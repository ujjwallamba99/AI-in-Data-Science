## Read(r)


```python
f = open("ujjwalfile.txt","r")
data = f.read()
print(data) #print data
print(type(data)) #print data type
f.close()
```

```python
f = open("ujjwalfile.txt","r")
data = f.read(50)             #takes length not indexing
print(data) #print data
print(type(data)) #print data type
f.close()
```

```python
f = open("ujjwalfile.txt","r")
line_1 = f.readline()
print(line_1)              #print data at particular line
line_2 = f.readline()
print(line_2)
line_3 = f.readline()
print(line_3)
line_4 = f.readline()
print(line_4)

data = f.read()
print(data) #print data which is left after execution of line
print(type(data)) #print data type
f.close()
```

```python
f = open("ujjwalfile.txt","r")
line_1 = f.readline().strip()     #first method        #strip is used to remove extra line from the output
print(line_1)
line_2 = f.readline().strip()
print(line_2)
line_3 = f.readline()
print(line_3)
line_4 = f.readline()
print(line_4)

data = f.read()
print(data) #print data which is left after execution of line
print(type(data)) #print data type
f.close()
```

```python
f = open("ujjwalfile.txt","r")
line_1 = f.readline()     #second method        #strip is used to remove extra line from the output
print(line_1,end = "")
line_2 = f.readline()
print(line_2)
line_3 = f.readline()
print(line_3)
line_4 = f.readline()
print(line_4)

data = f.read()
print(data) #print data which is left after execution of line
print(type(data)) #print data type
f.close()
```

```python
f = open("pythonfile.txt", "r")
data = f.read()
f.close()


if "python" in data.lower():    # Check if 'python' is in the text
    print("Found 'python' inside the file!")
else:
    print("'python' was not found.")
```

```python
# Use os import and use del keyword to delete from py. compiler.

import os

file_path = "pythonfile.txt"


if os.path.exists(file_path):    # Check if the file exists on your Mac before deleting
    os.remove(file_path)
    print(f"'{file_path}' has been deleted from your Mac storage.")
else:
    print(f"'{file_path}' does not exist.")
```

```python
#Check read lines(read line by line) and write lines(writes new line ,line by line)

with open("ujjwalfile.txt", "r") as f:
    line1 = f.readline()
    line2 = f.readline()

    print("Line 1:", line1.strip()) # .strip() removes extra newlines
    print("Line 2:", line2.strip())
```
