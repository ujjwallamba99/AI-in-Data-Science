##3.1 Line Chart

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
marks = [98,60,52,72,80]
noh = [10,6,5,7,8]

# Create a figure with a specific size (width, height)
plt.figure(figsize=(5,6))
# Plot the data with 'marks' on the x-axis and 'noh' on the y-axis.
# 'marker="s"' uses square markers and 'linestyle="-."' uses a dash-dot line style.
plt.plot(marks,noh,marker="s",linestyle="-.")

# Set the title of the plot
plt.title("Marks of Students",fontsize=14,color="blue")
# Set the label for the x-axis
plt.xlabel("Marks",fontsize=10,color="red")
# Set the label for the y-axis
plt.ylabel("No. of Study Hours",fontsize=10,color="green")


# Add a grid to the plot for better readability
plt.grid(True)
# Display the plot
plt.show()
```

```python
student={
    "marks":[20,22,25,30,35]
}
df= pd.DataFrame(student)

plt.figure(figsize=(6,5)) # Create a figure with a specific size (width, height)
plt.plot(df["marks"],marker="s",linestyle="-.") # Plot the 'marks' column from the DataFrame with square markers and a dash-dot line style

plt.title("Marks of Students",fontsize=14,color="blue")     # Set the title of the plot
plt.xlabel("Marks",fontsize=10,color="red") # Set the label for the x-axis
plt.ylabel("No. of Study Hours",fontsize=10,color="green") # Set the label for the y-axis


plt.grid(True) # Add a grid to the plot for better readability
plt.show() # Display the plot
```

```python
import pandas as pd
import matplotlib.pyplot as plt

student = {
    "marks": [20, 22, 25, 30, 35]
}

df = pd.DataFrame(student)

# FOR LOOP + RANGE + INDEX
print("Using for loop, range and index:")

for i in range(len(df)):
    print("Index:", i, "Marks:", df["marks"][i])


# ENUMERATE
print("\nUsing enumerate:")

for index, mark in enumerate(df["marks"]):
    print("Index:", index, "Marks:", mark)


#  WHILE LOOP
print("\nUsing while loop:")

i = 0

while i < len(df):
    print("Index:", i, "Marks:", df["marks"][i])
    i += 1


#  PLOTTING
plt.figure(figsize=(6, 5))

plt.plot(
    df.index,                 # Index addition
    df["marks"],
    marker="s",
    linestyle="-."
)

plt.title("Marks of Students", fontsize=14, color="blue")
plt.xlabel("Student Index", fontsize=10, color="red")
plt.ylabel("Marks", fontsize=10, color="green")

plt.grid(True)
plt.show()
```

```python
x = ["A","B","C","D","E"]
y = [10,6,5,7,8]

# Create a figure with a specific size (width, height)
plt.figure(figsize=(5,6))
# Plot the data with 'marks' on the x-axis and 'noh' on the y-axis.
# 'marker="s"' uses square markers and 'linestyle="-."' uses a dash-dot line style.
plt.plot(x,y,marker="s",linestyle="-.")


#Using for loop to print Text on points.
for i in range(len(x)):
  plt.text(x[i],y[i],str([x[i],y[i]]),ha="left",va = "top",fontsize=12)


# Set the title of the plot
plt.title("Marks of Students",fontsize=14,color="blue")
# Set the label for the x-axis
plt.xlabel("Marks",fontsize=10,color="red")
# Set the label for the y-axis
plt.ylabel("No. of Study Hours",fontsize=10,color="green")


# Add a grid to the plot for better readability
plt.grid(True)
# Display the plot
plt.show()
```

#3.2 Histogram

```python
import matplotlib.pyplot as plt
import pandas as pd
age = [-20,21,19,1,20,20,21,20,19,17,23,12,12,13,14,14,15,5,6,3,6]

plt.hist(age)

plt.title("Histogram",fontsize=14,color="blue")     # Set the title of the plot
plt.xlabel("Ages",fontsize=10,color="red") # Set the label for the x-axis
plt.ylabel("Frequency",fontsize=10,color="green") # Set the label for the y-axis


plt.grid(True) # Add a grid to the plot for better readability
plt.show() # Display the plot
```

```python
age = [10,20,20,20,20,22,22,30,25,45,50]

plt.figure(figsize=(6,5))

plt.hist(age,bins=4,edgecolor="k",color ="y",alpha=0.3) # Plotting a histogram with 4 bins, black edges, yellow bars, and 30% transparency

plt.title("Histogram",fontsize=14,color="blue")     # Set the title of the plot
plt.xlabel("Ages",fontsize=10,color="red") # Set the label for the x-axis
plt.ylabel("Frequency",fontsize=10,color="green") # Set the label for the y-axis


plt.grid(axis = "y") # Add a grid to the plot for better readability
plt.show() # Display the plot
```

Task

Create a line plot of the students' marks and perform the following:

1.Calculate the mean of the marks. \
2.Calculate the median of the marks.\
3.Plot the marks using plt.plot().\
4.Add a horizontal line for the mean using:

```python
#Add a horizontal line for the median using:
import pandas as pd
import matplotlib.pyplot as plt

# Data
student = {
    "marks": [20, 22, 25, 25, 28, 30, 32, 35, 40, 80]
}

df = pd.DataFrame(student)

# Calculate mean and median
mean = df["marks"].mean()
median = df["marks"].median()

print("Mean:", mean)
print("Median:", median)

# Histogram

plt.figure(figsize=(8, 5))

plt.hist(
    df["marks"],
    bins=5,
    edgecolor="black"
)

# Mean line
plt.axvline(
    mean,
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean:.2f}",color="y"
)

# Median line
plt.axvline(
    median,
    linestyle="-",
    linewidth=2,
    label=f"Median = {median:.2f}",color="y"
)

# Title and labels
plt.title("Distribution of Student Marks", fontsize=14)
plt.xlabel("Marks", fontsize=11)
plt.ylabel("Number of Students", fontsize=11)

# Grid
plt.grid(axis = "y", alpha=0.3)

# Legend
plt.legend()

# Show plot
plt.show()

```

Task\
**Q.have students of 10th class,\
  subject = ["Hindi","English","Maths","Science","Sst"]\
  score_2026=[40,45,47,48,50]\
  score_2025= [45,55,65,77,89]\
  score_2024=[98,99,95,88,80]\

  use plt.legend and plt.show and make a bar graph by comparing these with multiple categories.**

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

subject = ["Hindi","English","Maths","Science","Sst"]
score_2026=[40,45,47,48,50]
score_2025= [45,55,65,77,25]
score_2024=[98,99,95,88,80]


x = np.arange(len(subject))
width = 0.25
plt.figure(figsize=(8, 5))

#Using for loop to print Text on points.
for i in range(len(subject)):
  plt.text(x[i]-width,score_2026[i],str(score_2026[i]),ha="center",va = "bottom",fontsize=10)

  plt.text(x[i],score_2025[i],str(score_2026[i]),ha="center",va = "bottom",fontsize=10)

  plt.text(x[i]+width,score_2024[i],str(score_2026[i]),ha="center",va = "bottom",fontsize=10)

plt.bar(x-width,score_2026 ,width,label="2026")
plt.bar(x,score_2025 ,width,label="2025")
plt.bar(x+width,score_2024,width,label="2024" )



plt.title("10 Class Data")
plt.xlabel("Subjects")
plt.ylabel("Score")


plt.xticks(x, subject)
plt.legend(loc = "upper right")
plt.show()
```

#3.4 Pie Chart

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

subjects = ["Technical","English","GK","Aptitude"]
score = [50,20,20,10]
colours= ["violet","yellow","red","pink"]

plt.pie(score,labels = subjects,autopct="%1.2i%%",startangle=90 ,\
        explode = (0,0,0.2,0),colors = (colours),shadow=True)

plt.legend(loc="upper left")
plt.title("Neet score")
plt.show()
```

#3.5 Scatter Plot

```python
x1 = [2,4,6,8,30]
y1 = [1,3,5,7,9]


x2 = [12,14,16,18,130]
y2 = [11,13,15,17,19]

rang = ["red","blue","yellow","green","black"]
Size = [10,50,100,150,200]

plt.scatter(x1,y1,c = rang,s = Size,alpha = 0.7)
plt.scatter(x2,y2,c = rang,s = Size,alpha = 0.3)

plt.colorbar(label = "Itensity")
plt.legend()

plt.title("Scatter Plot")
plt.xlabel("x even")
plt.ylabel("y odd")
plt.show()
```
