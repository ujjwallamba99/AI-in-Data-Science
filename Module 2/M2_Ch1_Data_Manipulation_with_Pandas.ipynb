**2.2 Performing Arithmetic Operations in DF**

```python
var= pd.DataFrame({'A':[1,2,3,4,5], 'B':[6,7,8,9,10]})
var["C"]= var["A"]+var['B']
var["C"]= var["A"]-var['B']
var["C"]= var["A"]*var['B']
var["C"]= var["A"]/var['B']
var["C"]= var["A"]%var['B']

var
```

**2.3 Filtering in DF**

```python
var= pd.DataFrame({'A':[1,2,3,4,5], 'B':[6,7,8,9,10]})
var["C"]= var["A"]+var['B']
var["pyhton"]=var["C"]>10    #Filtering
var
```

**2.4 Handling Missing Values**

```python
import numpy as np
import pandas as pd
Sr={
    "Name": pd.Series(['ujjwal','shyam',np.nan,'bunny','honey']),
    "Roll_num": pd.Series([10,20,30,40,50]),
    "Marks": pd.Series([10,20,np.nan,40,50]),
    "Subject":pd.Series(["Hindi","Sans","Maths",np.nan,np.nan]),
    "City":pd.Series(["Meerut","Noida","Delhi",np.nan,"Gurgaon"])
}
DF= pd.DataFrame(Sr)
print(DF)
```

```python
DF.info()
```

```python
DF.isnull()
```

```python
DF.isnull().sum()
```

```python
print(DF.isnull().sum().sum())
```

```python
#remove null values
DF.dropna()
```

```python
DF.dropna(subset=["Marks"])
```

```python
DF["Marks"]=DF["Marks"].fillna(0)
DF
```

```python
DF["Marks"] = DF["Marks"].fillna(DF["Marks"].max())
DF
```
