#Step 7 — Clean gender

```python
df[df["gender"].isin([0, 1])]["gender"].value_counts()
```

#Step 7A — Standardize gender

```python
df["gender"] = df["gender"].replace({
    "1": "Male",
    "0": "Female",
    "M": "Male",
    "male": "Male",
    "F": "Female",
    "female": "Female"
})
```

```python
df["gender"].value_counts(dropna=False)
```

#Step 7C — Handle the 50 missing genders

```python
df["gender"] = df["gender"].fillna("Unknown")
```

```python
df["gender"].value_counts()
```

#Step 8 — Clean

```python
df["follow_up_required"].value_counts(dropna=False)
```

```python
#Conbvert error to yes and no
df["follow_up_required"] = df["follow_up_required"].replace({
    "0": "No",
    "Y": "Yes",
    "Yes": "Yes",
    "N": "No",
    "1": "Yes",
    "No": "No"
})
```

```python
df["follow_up_required"].value_counts(dropna=False)
```

#Step 9 — Clean billing_amount

#9A — Create a currency name and currency value column

```python
df["currency"] = df["billing_amount"].astype(str).str.extract(r"(^[A-Za-z€£$]+)")[0]
```

```python
df["currency"].value_counts(dropna=False)
```

#9B — Extract the numeric amount

```python
df["billing_numeric"] = (
    df["billing_amount"]
    .astype(str)
    .str.replace(r"[^0-9.]", "", regex=True)
)

df["billing_numeric"] = pd.to_numeric(
    df["billing_numeric"],
    errors="coerce"
)
```

```python
df[["billing_amount", "currency", "billing_numeric"]].head(10)
```
